```python
import nltk
import random
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
```

```python
nltk.download('punkt')
```

```python
vectorizer = TfidfVectorizer()
```

```python
import json
classifier = LogisticRegression(max_iter=10000,random_state=42)

# Load the JSON file
with open('intents.json', 'r') as file:
    data = json.load(file)

# Access the intent data
intents = data['intents']

tags = []
patterns = []

for intent in intents:
    for pattern in intent['patterns']:
        # Tokenize each word in the sentence
        patterns.append(pattern)
        tags.append(intent['tag'])

x = vectorizer.fit_transform(patterns)
y = tags

classifier.fit(x,y)
```

```python
def chatbot_response(text):
    # Predict the intent of the user's message
    input_text = vectorizer.transform([text])
    tag_data = classifier.predict(input_text)

    # Find the corresponding response
    for intent in intents:
        if intent['tag'] == tag_data[0]:
            return random.choice(intent['responses'])

    # If no match found (fallback)
    return "I'm sorry, I don't understand."
```

```python
chatbot_response("I need help")
```

```python
while True:
    query = input("User: ")
    output = chatbot_response(query)
    print("Chatbot-> {}".format(output))
```
